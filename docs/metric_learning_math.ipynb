# Metric Learning – Mathematical Formulation

This notebook explains the mathematical foundations of the metric-learning
components implemented in the `fashion_metric.metric_learning` and
`fashion_metric.evaluation` modules.

## Embedding function
- Learn a mapping $f_\theta : \mathcal{X} \to \mathbb{R}^d$ so that
  images from the same class are close in the embedding space.

## Triplet loss
For each triplet $(x_a, x_p, x_n)$ with labels $y_a = y_p \ne y_n$ and
embeddings $z_a = f_\theta(x_a)$, $z_p = f_\theta(x_p)$, $z_n = f_\theta(x_n)$,
the loss is
\begin{equation}
L = \max\big(0, \; \|z_a - z_p\|_2^2 - \|z_a - z_n\|_2^2 + \alpha\big),
\end{equation}
with margin $\alpha > 0$.

The implementation uses *semi-hard* and *hard* negatives as in
`InformativeNegativeTripletSelector`.

## Recall@k
For each query embedding we retrieve its $k$ nearest neighbours in cosine
similarity (equivalently L2-normalised dot product) and compute
\begin{equation}
\text{Recall@}k = \frac{1}{N} \sum_{i=1}^N \mathbb{1}\big[ \exists j \in N_k(i): y_j = y_i \big].
\end{equation}

## Error–coverage curve
Given a confidence score $c_i$ (average similarity of the top-$k$ matches)
and threshold $\tau$:
\begin{align}
\text{coverage}(\tau) &= \frac{\#\{i : c_i \ge \tau\}}{N},\\
\text{error}(\tau) &= \frac{\#\{i : c_i \ge \tau,\; \hat{y}_i \ne y_i\}}{\#\{i : c_i \ge \tau\}}.
\end{align}

The function `plot_error_coverage_curve` visualises this trade-off.


## Classification loss and metrics 
Cross entropy loss for predicted probabilities $p_{i,c}$ is
$$\mathcal{L}_{\text{CE}} = - \frac{1}{B} \sum_{i=1}^{B} \log p_{i, y_i}. $$
The standard accuracy is given by
$$\text{accuracy} = \frac{1}{N} \sum_{i=1}^{N} \mathbf{1}\{\hat{y}_i = y_i\}. $$
Balanced accuracy averages recall across classes and is computed as
$$\text{balanced accuracy} = \frac{1}{C} \sum_{c=1}^{C} \frac{1}{|D_c|} \sum_{(x_i,y_i) \in D_c} \mathbf{1}\{\hat{y}_i = c\}. $$
